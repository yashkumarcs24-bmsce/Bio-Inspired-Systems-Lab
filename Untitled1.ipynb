```python

import numpy as np
import random
import matplotlib.pyplot as plt

# ---------------- INPUT PARAMETERS ----------------

POP_SIZE = 100
GENERATIONS = 100

HEAD_LENGTH = 6
MAX_ARITY = 2
TAIL_LENGTH = HEAD_LENGTH * (MAX_ARITY - 1) + 1
CHROMOSOME_LENGTH = HEAD_LENGTH + TAIL_LENGTH

R_MIN = 0.05
R_MAX = 0.15

P_MIN = 0.05
P_MAX = 0.15

CROSSOVER_RATE = 0.2
ALPHA = 2.0

FUNCTIONS = ['+', '-', '*', '/']
TERMINALS = ['x', '1', '2']

random.seed(42)
np.random.seed(42)

# ---------------- TARGET FUNCTION ----------------

def target_function(x):
    return x**5 - 2*x**3 + x

X = np.linspace(-1, 1, 100)
Y = target_function(X)


# ---------------- CHROMOSOME GENERATION ----------------

def random_chromosome():
    head = [
        random.choice(FUNCTIONS + TERMINALS)
        for _ in range(HEAD_LENGTH)
    ]

    tail = [
        random.choice(TERMINALS)
        for _ in range(TAIL_LENGTH)
    ]

    return head + tail


# ---------------- EXPRESSION EVALUATION ----------------

def evaluate_expression(chromosome, x):

    position = [0]

    def decode():

        if position[0] >= len(chromosome):
            return np.ones_like(x)

        symbol = chromosome[position[0]]
        position[0] += 1

        if symbol == 'x':
            return x

        if symbol == '1':
            return np.ones_like(x)

        if symbol == '2':
            return np.full_like(x, 2.0)

        left = decode()
        right = decode()

        if symbol == '+':
            result = left + right

        elif symbol == '-':
            result = left - right

        elif symbol == '*':
            result = left * right

        elif symbol == '/':
            result = np.where(
                np.abs(right) > 1e-6,
                left / np.where(
                    np.abs(right) > 1e-6, right, 1
                ),
                left
            )

        else:
            result = np.ones_like(x)

        return np.clip(result, -1e6, 1e6)

    try:
        return decode()
    except (ValueError, OverflowError, FloatingPointError):
        return np.full_like(x, 1e6)


# ---------------- FITNESS FUNCTION ----------------

def calculate_fitness(chromosome):

    predicted = evaluate_expression(chromosome, X)

    if not np.all(np.isfinite(predicted)):
        return 0

    mse = np.mean((Y - predicted) ** 2)

    return 1 / (1 + mse)


# ---------------- FITNESS COEFFICIENT ----------------

def fitness_coefficient(fitness_values):

    average = np.mean(fitness_values)
    maximum = np.max(fitness_values)
    minimum = np.min(fitness_values)

    return (2 * average) / (maximum + minimum + 1e-10)


# ---------------- SELECTION ----------------

def select_parent(population, fitness_values):

    indices = random.sample(range(POP_SIZE), 3)

    best = max(indices, key=lambda i: fitness_values[i])

    return population[best].copy()


# ---------------- MUTATION ----------------

def mutate(chromosome, mutation_rate):

    chromosome = chromosome.copy()

    for i in range(CHROMOSOME_LENGTH):

        if random.random() < mutation_rate:

            if i < HEAD_LENGTH:
                chromosome[i] = random.choice(
                    FUNCTIONS + TERMINALS
                )
            else:
                chromosome[i] = random.choice(TERMINALS)

    return chromosome


# ---------------- RECOMBINATION ----------------

def recombine(parent1, parent2):

    child1 = parent1.copy()
    child2 = parent2.copy()

    if random.random() < CROSSOVER_RATE:

        point = random.randint(
            1, CHROMOSOME_LENGTH - 1
        )

        child1 = (
            parent1[:point] + parent2[point:]
        )

        child2 = (
            parent2[:point] + parent1[point:]
        )

    return child1, child2


# ---------------- INITIAL POPULATION ----------------

population = [
    random_chromosome()
    for _ in range(POP_SIZE)
]

best_chromosome = None
best_fitness = -1

previous_coefficient = None
mutation_rate = 0.10

fitness_history = []
mse_history = []


# ---------------- DGEP PROCESS ----------------

for generation in range(GENERATIONS):

    fitness_values = [
        calculate_fitness(chromosome)
        for chromosome in population
    ]

    # Save best chromosome
    best_index = np.argmax(fitness_values)

    if fitness_values[best_index] > best_fitness:

        best_fitness = fitness_values[best_index]

        best_chromosome = population[
            best_index
        ].copy()

    # Calculate population fitness coefficient
    current_coefficient = fitness_coefficient(
        fitness_values
    )

    # -------- DGEP-R: ADAPTIVE REGENERATION --------

    regeneration_rate = R_MIN + (
        R_MAX - R_MIN
    ) * np.exp(
        -ALPHA * current_coefficient
        * (generation + 1) / GENERATIONS
    )

    new_count = int(regeneration_rate * POP_SIZE)

    selected_count = POP_SIZE - new_count

    new_population = []

    # Generate new individuals
    for _ in range(new_count):

        new_population.append(
            random_chromosome()
        )

    # Select existing individuals
    for _ in range(selected_count):

        new_population.append(
            select_parent(population, fitness_values)
        )

    # -------- DGEP-M: DYNAMIC MUTATION --------

    if previous_coefficient is not None:

        if current_coefficient > previous_coefficient:

            mutation_rate = P_MIN

        elif np.isclose(
            current_coefficient, previous_coefficient
        ):

            mutation_rate = mutation_rate

        else:

            mutation_rate = min(
                (previous_coefficient /
                 (current_coefficient + 1e-10))
                * mutation_rate,
                P_MAX
            )

    # Apply mutation and recombination
    offspring = []

    random.shuffle(new_population)

    for i in range(0, POP_SIZE, 2):

        parent1 = new_population[i]

        parent2 = new_population[
            (i + 1) % POP_SIZE
        ]

        child1, child2 = recombine(
            parent1, parent2
        )

        child1 = mutate(child1, mutation_rate)
        child2 = mutate(child2, mutation_rate)

        offspring.append(child1)
        offspring.append(child2)

    population = offspring[:POP_SIZE]

    # Update previous coefficient
    previous_coefficient = current_coefficient

    # Store convergence values
    current_best_fitness = calculate_fitness(
        best_chromosome
    )

    fitness_history.append(current_best_fitness)

    mse_history.append(
        (1 / current_best_fitness) - 1
    )

    if (generation + 1) % 10 == 0:

        print(
            f"Generation {generation + 1}: "
            f"Best Fitness = {current_best_fitness:.6f}, "
            f"Mutation Rate = {mutation_rate:.4f}, "
            f"Regeneration Rate = {regeneration_rate:.4f}"
        )


# ---------------- FINAL RESULTS ----------------

predicted = evaluate_expression(
    best_chromosome, X
)

mse = np.mean((Y - predicted) ** 2)

ss_total = np.sum((Y - np.mean(Y)) ** 2)
ss_residual = np.sum((Y - predicted) ** 2)

r2 = 1 - (
    ss_residual / (ss_total + 1e-10)
)

print("\n========== DGEP RESULTS ==========")

print("Best Chromosome:", ''.join(best_chromosome))

print(f"Best Fitness: {best_fitness:.6f}")
print(f"Mean Squared Error: {mse:.6f}")
print(f"R2 Score: {r2:.6f}")

print("\nSample Predictions:")

for i in range(0, len(X), 10):

    print(
        f"x = {X[i]:.2f}, "
        f"Actual = {Y[i]:.4f}, "
        f"Predicted = {predicted[i]:.4f}"
    )


# ---------------- GRAPH 1: FITNESS CONVERGENCE ----------------

plt.figure(figsize=(9, 5))

plt.plot(
    range(1, GENERATIONS + 1),
    fitness_history,
    label="Best Fitness"
)

plt.xlabel("Generation")
plt.ylabel("Fitness")
plt.title("DGEP Fitness Convergence")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()


# ---------------- GRAPH 2: ACTUAL VS PREDICTED ----------------

plt.figure(figsize=(9, 5))

plt.plot(
    X, Y,
    label="Actual Function",
    linewidth=2
)

plt.plot(
    X, predicted,
    label="GEP Predicted Function",
    linestyle="--"
)

plt.xlabel("Input x")
plt.ylabel("Output y")
plt.title("Actual vs Predicted Function using DGEP")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()


# ---------------- GRAPH 3: MSE CONVERGENCE ----------------

plt.figure(figsize=(9, 5))

plt.plot(
    range(1, GENERATIONS + 1),
    mse_history,
    label="Best MSE"
)

plt.xlabel("Generation")
plt.ylabel("Mean Squared Error")
plt.title("DGEP Error Convergence")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
